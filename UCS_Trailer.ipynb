
# Practical Exercise 3: Uniform Cost Search

## 1. Problem Formulation

Using the highway network from Part 1, we apply Uniform Cost Search (UCS) to find the shortest route for the trailer from Nairobi to Kisumu.

- **State:** The town the trailer is currently in.
- **Start:** Nairobi.
- **Goal:** Kisumu.
- **Actions:** Travel from one town to a connected town.
- **Path cost:** Road distance in kilometres.
- **Search method:** Uniform Cost Search (UCS).

UCS explores the town with the lowest total distance from Nairobi first. The search continues until Kisumu is selected.



## 2. Directed Graph

The highway network is represented as a weighted directed graph. Each town is a node and each road connection has a distance in kilometres.

The graph uses the same towns and path costs from Part 1.


```python

graph = {
    "Nairobi": {"Thika": 45, "Maai Mahiu": 55, "Narok": 145},
    "Thika": {"Murang'a": 47},
    "Murang'a": {"Nyeri": 65},
    "Nyeri": {"Nyahururu": 104},
    "Nyahururu": {"Nakuru": 67},
    "Maai Mahiu": {"Naivasha": 34},
    "Naivasha": {"Gilgil": 30},
    "Gilgil": {"Nakuru": 42},
    "Nakuru": {"Molo": 47},
    "Molo": {"Londiani": 22},
    "Londiani": {"Kericho": 49},
    "Narok": {"Bomet": 81},
    "Bomet": {"Kericho": 76},
    "Kericho": {"Kisumu": 84},
    "Kisumu": {}
}

print("Highway network loaded successfully.")
print("Total towns:", len(graph))

```


## 3. Uniform Cost Search

UCS uses a priority queue to explore the path with the lowest total cost first.

We use `heapq` to manage the queue. Each time a town is selected, the algorithm checks its neighbouring towns and updates their costs if a shorter path is found.


```python

import heapq

def uniform_cost_search(graph, start, goal):
    # Queue stores (total cost, current town, path)
    queue = [(0, start, [start])]
    best_cost = {start: 0}
    step = 0

    while queue:
        cost, town, path = heapq.heappop(queue)

        # Ignore paths that are no longer the cheapest
        if cost > best_cost[town]:
            continue

        step += 1
        print(f"Step {step}: {town} ({cost} km)")

        # Stop when we reach Kisumu
        if town == goal:
            return path, cost

        # Check the roads leaving the current town
        for neighbour, distance in graph[town].items():
            new_cost = cost + distance

            if new_cost < best_cost.get(neighbour, float("inf")):
                best_cost[neighbour] = new_cost
                heapq.heappush(
                    queue,
                    (new_cost, neighbour, path + [neighbour])
                )

    return None, float("inf")
```


## 4. Search Traversal

The search starts from Nairobi and explores towns based on their cumulative distance. At each step, the town with the lowest cost is selected from the priority queue.


```python

start = "Nairobi"
goal = "Kisumu"

path, cost = uniform_cost_search(graph, start, goal)

print("\nOptimal Path:")
print(" → ".join(path))

print("\nTotal Distance:", cost, "km")
```


## 5. Optimal Path

Using Uniform Cost Search, the shortest path from Nairobi to Kisumu is:

**Nairobi → Maai Mahiu → Naivasha → Gilgil → Nakuru → Molo → Londiani → Kericho → Kisumu**

**Total distance: 363 km**

The algorithm explored the towns based on their cumulative path costs and selected Kisumu when it had the lowest cost in the priority queue.

Since all road distances are positive, UCS guarantees the shortest path within the directed graph used.
